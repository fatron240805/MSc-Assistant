```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import time
import os

# Cấu hình User-Agent để giả lập trình duyệt
HEADERS = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

# def download_html_to_excel(url_list, output_excel):
#     """Bước 1: Tải HTML từ danh sách URL nguồn và lưu vào Excel"""
#     data = []
#     print("--- Bước 1: Đang tải HTML nguồn ---")
#     for url in url_list:
#         try:
#             response = requests.get(url, headers=HEADERS, timeout=10)
#             if response.status_code == 200:
#                 data.append({'link': url, 'html_content': response.text})
#                 print(f"Tải thành công: {url}")
#             else:
#                 print(f"Lỗi {response.status_code} tại {url}")
#         except Exception as e:
#             print(f"Lỗi kết nối {url}: {e}")
#         time.sleep(1)
    
#     df = pd.DataFrame(data)
#     df.to_excel(output_excel, index=False)
#     print(f"Đã lưu file: {output_excel}")

# def extract_names_to_txt(input_excel, output_txt):
#     """Bước 2: Dùng logic trích xuất của bạn để lấy tên từ file Excel"""
#     print("\n--- Bước 2: Đang trích xuất individual ---")
#     df = pd.read_excel(input_excel)
#     all_names = []

#     for html_content in df['html_content']:
#         # SỬ DỤNG ĐÚNG LOGIC CỦA BẠN TẠI ĐÂY
#         soup = BeautifulSoup(html_content, 'html.parser')
#         character_links = soup.select('.mw-category-group ul li a')
        
#         for link in character_links:
#             name = link.get_text().strip()
#             if name:
#                 all_names.append(name)

#     # Ghi ra file txt, mỗi hàng một tên
#     with open(output_txt, 'w', encoding='utf-8') as f:
#         for name in all_names:
#             f.write(name + '\n')
    
#     print(f"Đã trích xuất {len(all_names)} tên vào file: {output_txt}")

def download_details_by_name(input_txt, output_final_excel):
    """Bước 3: Tải HTML chi tiết từng nhân vật dựa trên tên từ file txt"""
    print("\n--- Bước 3: Đang tải HTML chi tiết từng nhân vật ---")
    base_wiki_url = "https://vi.wikipedia.org/wiki/"
    final_results = []

    if not os.path.exists(input_txt):
        return

    with open(input_txt, 'r', encoding='utf-8') as f:
        names = [line.strip() for line in f if line.strip()]

    for name in names:
        # Quy tắc: tên miền + tên người, cách nhau bằng "_"
        formatted_name = name.replace(" ", "_")
        url = base_wiki_url + formatted_name
        
        print(f"Đang requests: {url}")
        try:
            response = requests.get(url, headers=HEADERS, timeout=10)
            if response.status_code == 200:
                final_results.append({
                    'individual': name, 
                    'html_content': response.text
                })
            else:
                print(f"Không tìm thấy bài viết cho: {name}")
        except Exception as e:
            print(f"Lỗi khi tải {name}: {e}")
        
        # Nghỉ ngắn để tránh bị chặn
        time.sleep(0.5)

    # Lưu kết quả cuối cùng vào Excel
    df_final = pd.DataFrame(final_results)
    df_final.to_excel(output_final_excel, index=False)
    print(f"\nHoàn tất! Kết quả cuối cùng tại: {output_final_excel}")

# --- CHẠY CHƯƠNG TRÌNH ---
if __name__ == "__main__":
    # URL danh sách các nhân vật (Trang Thể loại)
    
    # Khai báo tên file
    file_txt = 'danh_sach_nhan_vat.txt'
    file_ket_qua = "individual.xlsx"

    # Chạy quy trình
    download_details_by_name(file_txt, file_ket_qua)
```
