```python
import json
import re
from bs4 import BeautifulSoup

def process_hue_data(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, 'html.parser')
    
    for script_or_style in soup(["script", "style"]):
        script_or_style.decompose()
    all_text = soup.get_text(separator=' ', strip=True)

    data_struct = {
        "notable_info": {
            "title": "Sơ lược về chín chúa, mười ba vua triều Nguyễn",
            "summary": "Theo lịch sử, triều Nguyễn có 9 đời chúa và 13 đời vua. Chúa Tiên Nguyễn Hoàng là vị chúa Nguyễn đầu tiên và Bảo Đại là vị vua cuối cùng (1558 - 1945).",
            "nine_lords": [],
            "thirteen_kings": []
        },
        "all_content": all_text,
    }

    # Tách đoạn văn bản chứa danh sách Chúa và Vua
    # Dùng regex để tìm các đoạn bắt đầu bằng số thứ tự "1. Chúa..." hoặc "1. Vua..."
    pattern = r'(\d+)\.\s+((?:Chúa|Vua)\s+[^(\n]+)\s*\(([^)]+)\)\.\s*(.*?)(?=\s*\d+\.\s+(?:Chúa|Vua)|Mười ba vua|Nguồn:|$)'

    matches = re.findall(pattern, all_text, re.DOTALL)

    for match in matches:
        stt, name_raw, period, content = match
        
        full_name = name_raw.strip()
        
        item = {
            "stt": int(stt),
            "name": full_name,
            "period": period.strip(),
            "all_content": content.strip()
        }

        if "Chúa" in full_name:
            data_struct["notable_info"]["nine_lords"].append(item)
        else:
            data_struct["notable_info"]["thirteen_kings"].append(item)

    return data_struct

if __name__ == "__main__":
    try:
        final_data = process_hue_data('hue.txt')
        
        with open('hue.json', 'w', encoding='utf-8') as f:
            json.dump(final_data, f, ensure_ascii=False, indent=4)
            
        print("Đã trích xuất thành công!")
    except FileNotFoundError:
        print("Không tìm thấy file hue.txt.")
```
